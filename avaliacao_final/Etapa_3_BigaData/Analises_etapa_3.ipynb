```python
import pandas as pd

df = pd.read_csv("/content/avaliacao_transactions.csv", parse_dates=["timestamp"])

# KPIs gerais
total_transacoes = len(df)
total_fraudes = df["is_fraud"].sum()
taxa_fraude = total_fraudes / total_transacoes * 100

# Fraude por canal
analise_canal = (
    df.groupby("channel")
      .agg(
          total_transacoes=("transaction_id", "count"),
          total_fraudes=("is_fraud", "sum"),
          risco_medio=("risk_score", "mean"),
          valor_total=("amount", "sum")
      )
      .reset_index()
)

analise_canal["taxa_fraude"] = (
    analise_canal["total_fraudes"]
    / analise_canal["total_transacoes"]
    * 100
)

print(analise_canal)
```

# Analise feita por categoria  

```python
analise_categoria = (
    df.groupby("merchant_category")
      .agg(
          total_transacoes=("transaction_id", "count"),
          total_fraudes=("is_fraud", "sum"),
          ticket_medio=("amount", "mean"),
          risco_medio=("risk_score", "mean")
      )
      .reset_index()
)

analise_categoria["taxa_fraude"] = (
    analise_categoria["total_fraudes"]
    / analise_categoria["total_transacoes"]
    * 100
)
```

# Analise feita para segmento

```python
analise_segmento = (
    df.groupby("segment")
      .agg(
          total_transacoes=("transaction_id", "count"),
          total_fraudes=("is_fraud", "sum")
      )
      .reset_index()
)

analise_segmento["taxa_fraude"] = (
    analise_segmento["total_fraudes"]
    / analise_segmento["total_transacoes"]
    * 100
)
```

#Analise da pergunta de negocio elaborada pelo grupo

```python
analise_combinada = (
    df.groupby(["channel", "merchant_category"])
      .agg(
          total_transacoes=("transaction_id", "count"),
          total_fraudes=("is_fraud", "sum")
      )
      .reset_index()
)

analise_combinada["taxa_fraude"] = (
    analise_combinada["total_fraudes"]
    / analise_combinada["total_transacoes"]
    * 100
)

analise_combinada = analise_combinada.sort_values(
    "taxa_fraude",
    ascending=False
)
```

```python
import pandas as pd

# ============================================================
# ETAPA 3 - ANÁLISE DOS DADOS
# Disciplina: Big Data
# Alunos: Gabriel Ramos e Thales Vieira
# ============================================================

# Leitura da base
df = pd.read_csv(
    "avaliacao_transactions.csv",
    parse_dates=["timestamp"]
)

print("\n========== ETAPA 3 - ANÁLISE TECHPAY ==========")
print(f"Total de transações: {len(df)}")

# ------------------------------------------------------------
# 1. KPIs GERAIS
# ------------------------------------------------------------

total_fraudes = df["is_fraud"].sum()
taxa_fraude = (total_fraudes / len(df)) * 100

print("\n--- 1. KPIs GERAIS ---")
print(f"Total de transações: {len(df)}")
print(f"Total de fraudes: {total_fraudes}")
print(f"Taxa geral de fraude: {taxa_fraude:.2f}%")

# ------------------------------------------------------------
# 2. FRAUDE POR SEGMENTO
# ------------------------------------------------------------

segmento = df.groupby("segment").agg(
    transacoes=("transaction_id", "count"),
    fraudes=("is_fraud", "sum")
)

segmento["taxa_fraude"] = (
    segmento["fraudes"] / segmento["transacoes"] * 100
)

print("\n--- 2. FRAUDE POR SEGMENTO ---")
print(segmento.round(2))

# ------------------------------------------------------------
# 3. FRAUDE POR CANAL
# ------------------------------------------------------------

canal = df.groupby("channel").agg(
    transacoes=("transaction_id", "count"),
    fraudes=("is_fraud", "sum")
)

canal["taxa_fraude"] = (
    canal["fraudes"] / canal["transacoes"] * 100
)

print("\n--- 3. FRAUDE POR CANAL ---")
print(canal.round(2))

# ------------------------------------------------------------
# 4. FRAUDE POR CATEGORIA
# ------------------------------------------------------------

categoria = df.groupby("merchant_category").agg(
    transacoes=("transaction_id", "count"),
    fraudes=("is_fraud", "sum")
)

categoria["taxa_fraude"] = (
    categoria["fraudes"] / categoria["transacoes"] * 100
)

print("\n--- 4. FRAUDE POR CATEGORIA ---")
print(categoria.round(2))

# ------------------------------------------------------------
# 5. ANÁLISE TEMPORAL
# ------------------------------------------------------------

df["hora"] = df["timestamp"].dt.hour

horario = df.groupby("hora").agg(
    transacoes=("transaction_id", "count"),
    fraudes=("is_fraud", "sum")
)

horario["taxa_fraude"] = (
    horario["fraudes"] / horario["transacoes"] * 100
)

print("\n--- 5. FRAUDE POR HORA ---")
print(horario.round(2))

# ------------------------------------------------------------
# 6. PERGUNTA DE NEGÓCIO
# Canal + categoria
# ------------------------------------------------------------

combinacao = df.groupby(
    ["channel", "merchant_category"]
).agg(
    transacoes=("transaction_id", "count"),
    fraudes=("is_fraud", "sum")
)

combinacao["taxa_fraude"] = (
    combinacao["fraudes"] / combinacao["transacoes"] * 100
)

combinacao = combinacao.sort_values(
    "taxa_fraude",
    ascending=False
)

print("\n--- 6. CANAL + CATEGORIA ---")
print(combinacao.round(2))

print("\n========== FIM DA ANÁLISE ==========")
```
